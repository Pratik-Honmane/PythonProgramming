```python
# Operators = operators are used to perform operation on operands
# There are various types of operators

# 1) Comparison Operator (<,>,<=,>=,!=) = to compare the two conditions if both values are fulfill condition then it return the True output
# Less Than Operator(<) 
a = 10
b = 20

a < b  # return True because a is less than b
```

```python
b < a # return False because b is not less than a
```

```python
# Greater than Operator(>)

a = 200
b = 100

a > b # return True because a is greater than b
```

```python
# Equal to (==) => if both operands value is same then the condition returns True

a = 100
b = 300

a == b  # return False because a is not equal to b
```

```python
a = "hello"
b = "pratiik"

a == b  # because both sides not have same strings
```

```python
a = [10,20,30]
b = [10,20,30]
a == b  # Because a and b have same data and same data type
```

```python
a = [10,20,30]
b = (10,20,30)
a == b  # because a is list and b is tuple thats why it return False
```

```python
# Less than or equal to
a = 300
b = 300

a <= b  # a is less than b but a is not equal to b because the rule of 'OR' if both condition is FALSE then value become FALSE 
        # and if one is true and other is false then return value become TRUE
```

```python
a <= b
```

```python
# Greater than or equal to (>=)

a = 100
b = 200

a >= b
```

```python
b >= a
```

```python
a = 100
b = 100

a >= b
```

```python
a = [10,20,30]
b = [30,40,50]

a == b
```

```python
##################   Logical Operators   ####################
# Logical Operators are used when we combine two comparison operators together
# AND, OR AND NOT

a = 100
b = 200

a < b and b > a  # return True because both conditions are True
```

```python
len(str(a)) <= len(str(b)) and a < b
```

```python
# OR operator
'hello' == 'Hello' or 10 == 10
```
