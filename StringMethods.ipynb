```python
# String Operations

myString = "This is my string"
print(myString)

# Length Function
print("The length of string is : ", len(myString))

# Access character using index .
# Index always start from 0.
print(myString[0])
print(myString[5])
print(myString[16])

# Negative Index
print("We can access character from the string using negative index.")
print(myString[-1])

print(myString[len(myString)-1])

print(myString[17-1])
```

```python
# Slicing Operation = we get the substring from the main string or get range of characters from the string
myString = "This is my string"
print(myString)
```

```python
# Syntax = myString[start:end:step] # last character not included
myString[1:] # Get all string from the index 1
```

```python
myString[:5] # Get substring from 0 to 4
```

```python
myString[:-1] # last character not included
```

```python
myString[9:15] # Get character from index 9 to index 14 
```

```python
myString[-3:-1] # Negative index
```

```python
myString[-9:-1]
```

```python
myString[3:9:2]
```

```python
myString[::4]
```

```python
myString[1::4]
```

```python
myString[::0] # Slice step cannot be zero
```

```python
myString[::1] # Print whole string
```

```python
myString[::-1] # This last character to first character
```

```python
myString[::-3]
```

```python
# Upper() and lower() = to convert small text into capital and vice versa
print((myString).upper())
print((myString).lower())

# Another method convert small to capital
result = ''.join(map(str.upper,myString))
print(result)
```

```python
# Strip() = to remove white space from start and end
myString = " Hello, World! "
print(myString)

myString.lstrip()
myString.strip()
myString.rstrip()
myString.replace(" ","")

```

```python
# Split() = means split or divide the string using specified separator
print(myString)
myString.split(",")
```

```python
myString.split(" ")
```

```python
# Concatenation of String = Joining of strings
str1 = "Hello,"
str2 = "Pratik"
print(str1+ " " + str2)
```

```python
# Format String = f-string
age = 24
print(f"This is my age",age)
print(f"This is my age {age}")
```

```python
# Strings are immutable = we cannot change the string or update the string
name = "Pratik"
name[5] = 'i'
print(name)
```

```python
# Overcome this drawback
print(name[:4] + "iik")
```

```python
# Format String = We want to combine string and numbers together.
print("My name is {} and my age is {}".format("Pratiik",24))
```
