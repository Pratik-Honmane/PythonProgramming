```python
# Dictionary = Its is collection of objects that store in key:value pair.
# Dictionary can be declare using {}
# Dictionary can be unordered and mutable and repetition not allowed

# Create a dictionary

student_info = {
    "student_name":"Pratik Honmane",
    "student_education":"BE",
    "student_mobile":7517330553,
    "student_hobbies":["Drawing","Making Rangoli","Excercising","Running"],
}

print(student_info)
```

```python
# Access Dictionary Item using keys

student_info['student_hobbies']
```

```python
student_info['student_name']
```

```python
# If I want to access Excercising value from student_hobbies key
print(student_info['student_hobbies'][2])
```

```python
# Length of Dictionary
len(student_info)
```

```python
# data type
print(type(student_info))
```

```python
# Add object in the dictionary
student_info.update({"student_dob":"30/09/2001"})
```

```python
print(student_info)
```

```python
# Remove object from the dictionary
student_info.pop('student_dob')
```
