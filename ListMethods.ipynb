```python
# List = Collections of items of various data types
# List = ordered and mutable
# List = We can add or remove items or objects from the list

student_list = ["Pratiik","Adeeba","Farhan","Rushikesh"]
print(student_list)
```

```python
# check data type using type()
print(type(student_list))
```

```python
# length of list = use len()
print("The length of student list is : ", len(student_list))
```

```python
# Accessing list items = use indexing
print(student_list[1])
print(student_list[0])
print(student_list[2])
print(student_list[3])
```

```python
student_list
```

```python
teacher_list = ["Nitin Sir", "Ashish Sir"]
print(teacher_list)
```

```python
# Adding both list = use concatenation operator '+' Operator
newlist = student_list + teacher_list
print(newlist)
```

```python
# insert teacher in the newlist use = append() method
newlist.append('Dipti Mam') # add object at the end of the list
print(newlist)
```

```python
# remove item from the list use = pop() or remove() method
newlist.pop(7)
```

```python
print(newlist)
```

```python
# using remove method
newlist.remove('Dipti Mam')
```

```python
print(newlist)
```

```python
# Arrange item ascending or descending order use = sort() and reverse()
newlist.sort()
print(newlist)
```

```python
# Reverse list
newlist.reverse()
print(newlist)
```

```python
# Copy list into current list use = copy() method
print(teacher_list)
print(student_list)
```

```python
newlist1 = student_list.copy()
```
