```python
# Tuple = It is ordered and unchangeable items and cannot be add after the tuple is created
# Tuple create with round brackets ()
# Tuples are similar to lists however the key difference is "immutability".

# Create a tuple
my_hobbies = ("Running","Making Rangoli","Drawing","Exercising")
```

```python
# Printing tuple
print(my_hobbies)
```

```python
# length of tuple = use len() function
len(my_hobbies)
```

```python
# Count the occurrence of '2' in the my_new_tuple
my_new_tuple = (2,4,6,8,2,6,1,9)
print(my_new_tuple.count(2))  # 2 occure two times in the tuple
```

```python
# Return the index of items in the tuple
print(my_new_tuple.index(2))
```

```python
# Access the tuple items using index
my_new_tuple[0]
```

```python
my_new_tuple[-1]
```

```python
my_new_tuple[2:7:2]
```

```python
type(my_new_tuple[2:7:2])  # return type is tuple
```

```python
print(my_new_tuple)
```

```python
my_new_tuple[1] = 10  # error due to immutability we can not update after create
```

```python
my_new_tuple[::-1]  # make tuple reverse
```
