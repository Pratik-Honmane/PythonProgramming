```python
# Check Even Odd from the list

def even_odd_function(num_list):
    # placeholder variable
    even_number = []
    odd_number = []
    for num in num_list:
        if num % 2 == 0:
            even_number.append(num)
        else:
            odd_number.append(num)
    
    return even_number
            
```

```python
even_odd_function([1,2,3,4,5])
```

```python
# Factorial Number
def factorial_number(number):
    if number == 0:
        return 1
    else:
        return number * factorial_number(number-1)

factorial_number(5)
```

```python
# Reverse a string
def reverse_list(list1):
    return(list1[::-1])

result = reverse_list([10,20,30,40,50])
```

```python
# Check Palindrome = means when we read from start as well as reverse it should be same
# for example = madam

def palindrome_function(name):
    lower = name.lower()
    if lower == lower[::-1]:
        print("Given name is Palindrome")
    else:
        pass
```

```python
palindrome_function("Malayalam")
```

```python
# Count vowels in string
def count_vowels(name):
    vowels = "aeiou"
    count = 0
    for char in name:
        if char in vowels:
            count = count + 1
    return count
```

```python
count_vowels("Abhijeet")
```
